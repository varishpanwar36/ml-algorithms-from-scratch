```python
import numpy as np
from sklearn.datasets import load_diabetes

from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score
from sklearn.model_selection import train_test_split
import random

X,y = load_diabetes(return_X_y=True)
```

```python
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.2,random_state=2)
```

```python
reg=LinearRegression()
reg.fit(X_train,y_train)
```

```python
print(reg.coef_)
print(reg.intercept_)
```

```python
y_pred=reg.predict(X_test)
r2_score(y_test,y_pred)
```

```python
class MBGDRegressor:

  def __init__(self,learning_rate=0.01,epochs=100,batch_size=10):
    self.coef_=None;
    self.intercept_=None;
    self.lr=learning_rate;
    self.epochs=epochs;
    self.batch_size=batch_size;

  def fit(self,X_train,y_train):
    self.intercept_=0;
    self.coef_=np.ones(X_train.shape[1])

    for i in range(self.epochs):

      for j in range(int(X_train.shape[0]//self.batch_size)):
        #Generates random batch size indexes
        idx=random.sample(range(X_train.shape[0]),self.batch_size)

        y_hat=np.dot(X_train[idx],self.coef_)+ self.intercept_

        coef_der= -2*np.dot((y_train[idx]-y_hat),X_train[idx])/self.batch_size
        intercept_der=-2*np.mean(y_train[idx]-y_hat)

        self.coef_=self.coef_- (self.lr*coef_der)
        self.intercept_=self.intercept_ -(self.lr*intercept_der)
    print(self.intercept_,self.coef_)

  def predict(self,X_test):
    return np.dot(X_test,self.coef_)+self.intercept_
```

```python
mbgd=MBGDRegressor(learning_rate=0.3,epochs=100,batch_size=int(X_train.shape[0])//5)
```

```python
mbgd.fit(X_train,y_train)
```

```python
y_pred=mbgd.predict(X_test)
```

```python
r2_score(y_test,y_pred)
```

Mini Batch Implementation using SKlearn

```python
from sklearn.linear_model import SGDRegressor
```

```python
sgd=SGDRegressor(learning_rate='constant',eta0=0.2)
#We can't apply mini batch through sklearn
#There is a function called partial fit which allows to send a particular row in sgd regressor so we use that
```

```python
batch_size=35
for i in range(100):#100 means epochs=100
  idx=random.sample(range(X_train.shape[0]),batch_size)
  sgd.partial_fit(X_train[idx],y_train[idx])
```

```python
sgd.coef_
```

```python
sgd.intercept_
```

```python
y_pred=sgd.predict(X_test)
```

```python
r2_score(y_test,y_pred)
```
